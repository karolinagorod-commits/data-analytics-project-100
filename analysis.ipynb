```python
import requests
import pandas as pd

url_visits = " https://data-charts-api.hexlet.app/visits?begin=2023-03-01&end=2023-09-01"
url_regs = "https://data-charts-api.hexlet.app/registrations?begin=2023-03-01&end=2023-09-01"

response_visits = requests.get(url_visits)
response_regs = requests.get(url_regs)

data_visits = response_visits.json()
data_regs = response_regs.json()

df_api_visits = pd.DataFrame(data_visits)
df_api_regs = pd.DataFrame(data_regs)

print(f"Из API успешно загружено визитов: {len(df_api_visits)}")
print(f"Из API успешно загружено регистраций: {len(df_api_regs)}")
```
